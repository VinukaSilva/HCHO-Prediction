```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
import pickle
import numpy as np
from scipy import optimize
from folium.plugins import MarkerCluster,HeatMap
import matplotlib.cm as cm

from statsmodels.tsa.seasonal import seasonal_decompose
import warnings
warnings.filterwarnings('ignore')
```

```python
# Reading the pre processed dataset
hcho_data = pd.read_csv('preprocessed_dataset/hcho_data.csv')
hcho_data
```

```python
hcho_data.set_index('Location', inplace=True)
```

```python
# Assuming 'Next Date' column is not in datetime format
hcho_data['Next Date'] = pd.to_datetime(hcho_data['Next Date'])

# Filter data for the year 2019
hcho_data_2019 = hcho_data[hcho_data['Next Date'].dt.year == 2019]

# Group data by region
grouped = hcho_data_2019.groupby("Location")

# Create a Plotly figure
fig = go.Figure()

# Plot data for each region
for region, data in grouped:
    fig.add_trace(go.Scatter(x=data['Next Date'], y=data['HCHO Reading'], mode='lines', name=region))

# Adding labels and title
fig.update_layout(title='HCHO Readings of All Regions Over the Year 2019',
                  xaxis_title='Date',
                  yaxis_title='HCHO Reading',
                  xaxis=dict(
                      rangeselector=dict(
                          buttons=list([
                              dict(count=1, label="1m", step="month", stepmode="backward"),
                              dict(count=6, label="6m", step="month", stepmode="backward"),
                              dict(count=1, label="1y", step="year", stepmode="backward"),
                              dict(step="all")
                          ])
                      ),
                      rangeslider=dict(visible=True),
                      type="date"
                  )
                  )

# Display the plot
fig.show()
```

```python
# Assuming 'Next Date' column is not in datetime format
hcho_data['Next Date'] = pd.to_datetime(hcho_data['Next Date'])

# Filter data for the year 2019
hcho_data_2019 = hcho_data[hcho_data['Next Date'].dt.year == 2019]

# Extract month from 'Next Date' column
hcho_data_2019['Month'] = hcho_data_2019['Next Date'].dt.month

# Group data by month and sum HCHO readings
monthly_sum = hcho_data_2019.groupby('Month')['HCHO Reading'].sum().reset_index()

# Create a bar plot using Seaborn
plt.figure(figsize=(9, 5.5))
sns.barplot(x='Month', y='HCHO Reading', data=monthly_sum, palette='pastel')

# Adding labels and title
plt.title('Total HCHO Readings per Month in 2019 in all regions')
plt.xlabel('Month')
plt.ylabel('Total HCHO Reading')

# Display the plot
plt.tight_layout()
plt.show()
```

```python
weekly_sum = hcho_data["HCHO Reading"].resample("1W").sum()

# Plotting using Seaborn
plt.figure(figsize=(12, 5))
sns.barplot(x=weekly_sum.index.strftime('%Y-%m'), y=weekly_sum.values, palette='Paired')
plt.xlabel("Week")
plt.ylabel("Sum of HCHO rolling")
plt.title("Weekly Sum of HCHO emission rates (range) in All Locations")
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```
